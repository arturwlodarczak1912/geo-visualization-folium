```python
df[['lat', 'lng']].mean()
```

```python
m = folium.Map(location=df[['lat', 'lng']].mean())
m
```

Zadanie: Wyrysować mapę dla lokalizacji z naszych danych.

Zadanie: Pokazać na mapie dwa wybrane punkty z naszych danych

```python
m = folium.Map(location=(df_amsterdam['lat'].mean(), df_amsterdam['lng'].mean()))
folium.Marker(
    location=[df_amsterdam.loc[0,'lat'],df_amsterdam.loc[0,'lng']],
    tooltip="Click me!",
    popup="1",
    icon=folium.Icon(icon="cloud"),
).add_to(m)

folium.Marker(
    location=[df_amsterdam.loc[1,'lat'],df_amsterdam.loc[1,'lng']],
    tooltip="Click me!",
    popup="2",
    icon=folium.Icon(color="green"),
).add_to(m)
m
```

Zadanie: Zmienić tooltip, aby był rating.

```python
room_type = df.loc[0]['room_type']
guest_satisfaction_overall = df.loc[0]['guest_satisfaction_overall']
desc = f'{room_type}\n{guest_satisfaction_overall}'
desc
```

```python
amsterdam_location = df.loc[0][['lat','lng']]
m = folium.Map(location=amsterdam_location)

folium.Marker(
    location=amsterdam_location,
    tooltip=f"Price: {df.loc[0]['realSum']}",
    popup=desc,
    icon=folium.Icon(color='green')
  ).add_to(m)
m
```

```python
m = folium.Map(location=(df_amsterdam['lat'].mean(), df_amsterdam['lng'].mean()), zoom_start=12)
grp1 = folium.FeatureGroup("Ocena = 100").add_to(m)
grp2 = folium.FeatureGroup("Ocena 95-99").add_to(m)
grp3 = folium.FeatureGroup("Ocena = 90-94").add_to(m)
folium.LayerControl().add_to(m)

for i in range(len(df_amsterdam)):
  if df_amsterdam.loc[i,'guest_satisfaction_overall'] == 100:
    layer = grp1
  elif df_amsterdam.loc[i,'guest_satisfaction_overall'] >= 95:
    layer = grp2
  else:
    layer = grp3
  folium.Marker(
      location=[df_amsterdam.loc[i,'lat'],df_amsterdam.loc[i,'lng']],
      tooltip = df_amsterdam.loc[i,'guest_satisfaction_overall'],
      popup=f"Odległość od metra: ,<b>{round(df_amsterdam.loc[i,'metro_dist'],2)}</b>",
      icon=folium.Icon(icon="cloud"),
  ).add_to(layer)
m.save('index.html')
m
```
